```python
# Put all column name in SnakeCasing i.e all lower case or upper case
df.columns = df.columns.str.lower()
df.columns =  df.columns.str.replace(' ','_')
```

```python
df.columns
```

**4. Data Cleaning**

```python
# Drop unnecessary column from the dataset
df = df.drop('online/offline', axis=1)
```

```python
# Filling the missing values in the dataset
df['online_store'] = df['online_store'].fillna('In-store purchase')
df['size'] = df.groupby('item_purchased')['size'].transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else 'N/A'))
df['festival/sale'] = df['festival/sale'].fillna('Standard Day')
df['delivery_speed'] = df['delivery_speed'].fillna('Offline purchase')
```

```python
# create a new column age_group using feature engineering
labels = ['Young Adult', 'Adult','Middle Aged', 'Senior']
bins = [1, 18, 35, 60, 100]
df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels, right=True)
```

```python
df[['age','age_group']].head(10)
```

```python
df.isnull().sum()
```

**5. Feature Engineering**

```python
# create a new column age_group using feature engineering
labels = ['Young Adult', 'Adult','Middle Aged', 'Senior']
bins = [1, 18, 35, 60, 100]
df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels, right=True)
```

```python
df[['age','age_group']].head(10)
```

```python
df['frequency_of_purchases'].unique()
```

```python
# Create column purchase_frequency_days
frequency_mapping={
    'Weekly': 7,
    'Fortnightly': 14,
    'Monthly': 30,
    'Quarterly': 90,
    'Rarely': 365

}
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df[['purchase_frequency_days', 'frequency_of_purchases']].head(10)
```

```python
# Save the cleaned dataset to a new CSV file
df.to_csv('cleaned_customer_shopping_behavior.csv', index = False)
```
