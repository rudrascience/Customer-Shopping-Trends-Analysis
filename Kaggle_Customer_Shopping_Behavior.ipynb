```python
import pandas as pd
df = pd.read_csv('/content/customer_shopping_behavior.csv')
```

```python
df['Online Store'].fillna('In-store purchase', inplace=True)
df['Size'] = df.groupby('Item Purchased')['Size'].transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else 'N/A'))
df['Festival/Sale'].fillna('Standard Day', inplace=True)
df['Delivery Speed'].fillna('N/A(Offline purchase)', inplace=True)
```

```python
df.isnull().sum()
```

```python
# Put all column name in SnakeCasing i.e all lower case or upper case
df.columns = df.columns.str.lower()
df.columns =  df.columns.str.replace(' ','_')
```

```python
df.columns
```

```python
# create a new column age_group using feature engineering
labels = ['Young Adult', 'Adult', 'Senior']
bins = [18, 30, 60, 100]
df['age_group'] = pd.cut(df['age'], bins=bins, labels=labels, right=True)
```
